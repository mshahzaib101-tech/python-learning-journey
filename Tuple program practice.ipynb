# Sum & Average

```python
marks=(56,67,87,78,92,82)
sum1=0
for X in marks:
    sum1=sum1+X
print("Total marks:", sum1)
average=sum1/len(marks)
print("Average marks:", average)
```

# Convert tuple into list & list into tuple

```python
num=(10,20,30,40,50,60)
ls=list(num)
ls.append(70)
tp=tuple(ls)
print(tp)
```

```python
num=[10,20,30,40,50,60]
tp=tuple(num)
print(tp)
```

```python
num=(10,20,30,40,50,60)
ls=list(num)
print(ls)
```

```python
X=(10,20,30,40,50,60)
Y=(7,12,45)
ls1=list(X)
ls2=list(Y)
ls1.extend(ls2)
print("List:", ls1)
for Z in ls1:
  if(Z%2==0):
     print("Even:", Z)
tp=tuple(ls1)
print("Tuple:", tp)
```
