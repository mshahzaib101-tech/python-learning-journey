# Maximum Number

```python
numbers = [10, 45, 20, 80, 35]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("Largest:", largest)
```

# Minimum Number

```python
numbers = [10, 45, 20, 80, 35]

smallest = numbers[0]

for num in numbers:
    if num < smallest:
        smallest = num

print("Smallest:", smallest)
```

```python
numbers = [10, 20, 30, 40, 50]

total = 0

for num in numbers:
    total = total + num

average = total / len(numbers)

print("Total:", total)
print("Average:", average)
```

# Student Marks Management Example:

```python
marks=[56,76,65,90,89]
marks.sort(reverse=True)
total=0
largest=marks[0]
smallest=marks[0]
count=0
for X in marks:
    total=total+X
    if X>largest:
        largest=X
    if (X<smallest):
        smallest=X
        if (X>50):
            count=count+1
average=total/len(marks)
print("Total marks:", total)
print("Average marks:", average)
print("Largest marks:", largest)
print("Smallest marks:", smallest)
print(count)
print("Sorted:", marks)
```
