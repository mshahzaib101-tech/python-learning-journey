# 2. Strings are Sequences

* # Positive Indexing

```python
name = "PYTHON"

print(name[0])
print(name[1])
print(name[2])
```

* # Negative Indexing

```python
name = "PYTHON"

print(name[-1])
print(name[-2])
print(name[-3])
```

* # String Slicing 

```python
name = "PYTHON"
print(name[2:5])
```

```python
name = "PYTHON"

print(name[0:3])
print(name[2:5])
print(name[1:4])
```

```python
name = "PYTHON"

print(name[0:6:2])
```

* # Reverse a String Using Slicing 

```python
message = "Welcome to PYTHON"
print(message[:: -1])
```

```python
name = "PYTHON"
print(name[:: -1])
```

* # Strings are Immutable 

```python
name = "PYTHON"
name[0]="j"
print(name)
```

```python
name = "PYTHON"

name = "J" + name[1:]

print(name)
```

# 3. String Methods

1. # upper() 

```python
name = "shahzaib"

print(name.upper())
```

```python
name="welcome to the python"
print(name.upper())
```

2. # lower() 

```python
name="SHAHZAIB"
print(name.lower())
```

```python
name="WELCOME TO THE PYTHON"
print(name.lower())
```

3. # capitalize() 

```python
name = "shahzaib"

print(name.capitalize())
```

```python
name="welcome to the python"
print(name.capitalize())
```

4. # title() 

```python
name = "muhammad shahzaib"

print(name.title())
```

```python
name="welcome to the python"
print(name.title())
```

5. # strip() 

```python
name = "   'shahzaib'   "

print(name.strip())
```

```python
name="    'welcome to the python'      "
print(name.strip())
```

6. # replace() 

```python
message = "I like Java"

print(message.replace("Java", "Python"))
```

```python
message = "My name is ch"

print(message.replace("ch", "Shahzaib"))
```

7. # split() 

```python
message = "I love Python"

print(message.split())
```

```python
message = "My name is Shahzaib"

print(message.split())
```

8. # join()  

```python
words = ["I", "love", "Python"]

result = " ".join(words)

print(result)
```

```python
words = ['My', 'name', 'is', 'Shahzaib']

result = " ".join(words)

print(result)
```
