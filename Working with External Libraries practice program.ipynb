# *Practical Examples*

```python
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

# Check Data Type

print("Data Type:", type(numbers))


# Calculate Average

average = numbers.mean()

print("Average:", average)


# Find Maximum

maximum = numbers.max()

print("Maximum:", maximum)


# Find Minimum

minimum = numbers.min()

print("Minimum:", minimum)


# Convert Array into List

result = numbers.tolist()

print("Python List:", result)
```

```python
import numpy as np

numbers = np.array([5, 10, 15, 20, 25])

# Check Object Type

print("Object Type:")

print(type(numbers))


# Check Available Methods

print("Mean Method Available:")

print("mean" in dir(numbers))


# Calculate Average

average = numbers.mean()

print("Average:", average)


# Display Documentation

help(numbers.mean)
```
