# Maximum & Minimum

```python
numbers = [10, 25, 5, 80, 45, 60]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("Largest Number:", largest)
```

```python
numbers = [10, 25, 5, 80, 45, 60]

smallest = numbers[0]

for num in numbers:
    if num < smallest:
        smallest = num

print("smallest Number:", smallest)
```

```python
words = ["Python", "Java", "Programming", "AI", "Computer"]

lengths = []

for word in words:

    word_length = len(word)

    lengths.append(word_length)

print("Original Words:", words)
print("Word Lengths:", lengths)
```

# Advance Program

```python
numbers = list(range(1, 101))

for num in numbers:

    if num % 3 == 0 and num % 5 == 0:

          print("Divisible by 3 and 5:", num)
```

```python
for i in range(1, 6):

    for j in range(1, 6):

        result = i * j

        print(result, end="\t")

    print()
```

```python
for i in range(1, 6):

    for j in range(1, 6):

        print("*", end="\t")

    print()
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

result = []

for num in numbers:

    if num % 2 == 0:

        result.append("Even")

    else:

        result.append("Odd")

print("Original List:", numbers)
print("Result:", result)
```

```python
list1 = [10, 20, 30, 40, 50]

list2 = [30, 40, 50, 60, 70]

common = []

for num in list1:

    for value in list2:

        if num == value:

            common.append(num)

print("Common Elements:", common)
```

```python
prime_numbers = []

for num in range(2, 51):

    is_prime = True

    for i in range(2, num):

        if num % i == 0:

            is_prime = False

            break

    if is_prime == True:

        prime_numbers.append(num)

print("Prime Numbers:", prime_numbers)
```
