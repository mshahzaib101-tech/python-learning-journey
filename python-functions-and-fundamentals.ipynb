# Practical Programming Practice

```python
student_name = input("Enter Student Name: ")

marks = [75, 80, 65, 90, 85]

total = sum(marks)

maximum = max(marks)

minimum = min(marks)

average = total / len(marks)

print("Student Name:", student_name)

print("Total Marks:", total)

print("Maximum Marks:", maximum)

print("Minimum Marks:", minimum)

print("Average Marks:", round(average, 2))
```

# Functions Applied to Functions

```python
def square(x):
    return x * x

print(square(5))
```

```python
def double(x):
    return x * 2

def calculate(function, number):
    return function(number)

result = calculate(double, 5)

print(result)
```

```python
def mult_by_five(x):
    return 5 * x

def call(fn, arg):
    return fn(arg)

def squared_call(fn, arg):
    return fn(fn(arg))

print(call(mult_by_five, 1))

print(squared_call(mult_by_five, 1))
```

```python
def mod_5(x):
    return x % 5

result = max(100, 51, 14, key=mod_5)

print(result)
```
